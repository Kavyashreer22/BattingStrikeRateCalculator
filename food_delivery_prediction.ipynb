```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
df=pd.read_csv("food_delivery_prediction (1).csv")
print(df)
x=df[["Restaurant_Distance_km","Order_Time_Hour","Traffic_Level"]]
y=df["Delivery_Time_Min"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=11)
model=LinearRegression()
model.fit(x_train,y_train)
model.predict(x_test)
y_pred=model.predict(x_test)
print(y_pred)
print("Mean Squared Error", mean_squared_error(y_test,y_pred))
print("r2 Score",r2_score(y_test,y_pred))
res_km=float(input("Enter distance:"))
order_time=int(input("Enter order time:"))
traffic_level=int(input("Enter traffic level: "))
predicted_time=model.predict([[res_km,order_time,traffic_level]])
print(predicted_time)
plt.scatter(y_pred,y_test)
plt.xlabel("Predicted Time")
plt.ylabel("Actual Time")
plt.title("Delivery Time Prediction")
plt.show()

```
